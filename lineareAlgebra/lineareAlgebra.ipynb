####  A2a

```python
a = [2,1,-2,-1,-2,2,3,0,-1]
b = [-2,9,7]
A0 = np.array(a,dtype='float').reshape((3,3))
b = np.array(b).reshape((3,1))
print(A0)
```

```python
# Augmented Matrix bilden
A = np.hstack((A0,b))
print(A)
```

```python
# Elimination
B = np.hstack((A,b))
A[1] = 2*A[1]+A[0]
A[2] = 2*A[2]-3*A[0]
print(A)
```

```python
A[2] = A[2]-A[1]
print(A)
```

Die Stufenform (Echelon-Form) ist nicht eindeutig. Aber die Lösung bleibt die gleiche.

```python
np.linalg.solve(A[:,:3], A[:,3:])
```

```python
A = np.hstack((A0,b))
A = Matrix(A).echelon_form()
A
```

```python
sympy.linsolve([A[:,:3], A[:,3:]])
```

```python
# Rücksubstitution
# x1, x2, x3 sind zunächst Symbole, nach der Lösung dann Zahlen

x1, x2, x3 = symbols("x1 x2 x3")
x3 = solve(2*x3-4, x3)[0]
x2 = solve(-3*x2 +2*x3-16,x2)[0]
x1 = solve(2*x1+x2-2*x3+2)[0]
x = np.array([x1,x2,x3]).reshape((3,1))
print(x)
```

```python
# Kontrolle
print(A0 @ x)
```

```python
# Alles mit einem Befehl
np.linalg.solve(A0,b)
```

```python
# oder mit sympy

sympy.linsolve([Matrix(A0),Matrix(b)])
```
